```python
num = int(input("정수를 입력하세요: "))

if num > 0:
    print("양수입니다.")
elif num < 0:
    print("음수입니다.")
else:
    print("0입니다.")
```

```python
count = int(input("몇 개의 숫자를 입력하시겠습니까? "))
total = 0

for i in range(count):
    num = int(input(f"{i+1}번째 숫자를 입력하세요: "))
    total += num

average = total / count
print(f"합계: {total}")
print(f"평균: {average}")
```

```python
dan = int(input("출력할 단을 입력하세요 (예: 2, 3, 7...): "))

for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
```
