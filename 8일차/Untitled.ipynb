```python
while True:
    num = int(input("양의 정수를 입력(종료는 -1):"))
    if num == -1:
        break
    i = factorial = 1
    while i <= num:
        factorial = factorial * i
        i = i + 1
    print("%s!=%s" % (num, factorial))
```
