```python
A=[23, 15, 2, 14, 14, 16, 20, 14]
B=[2, 48, 15, 14, 26, 32, 47, 14]
C=[]
for item in A:
    if item in B:
        C.append(item)
print(C)
```

```python
A=[23, 15, 2, 14, 14, 16, 20, 14]
B=[2, 48, 15, 14, 26, 32, 47, 14]
D=[]
for item in A:
    if item not in B:
        D.append(item)
print(D) #in 연산을 이용한 for문으로 구현한 차집합
```
