```python
a=0
b=0
c=0
d=0
f=0

for i in range(5):
    score = int(input(str(i+1)+"번째 학생 점수:"))

    if score>=90:
        a+=1
    elif score>=80:
        b+=1
    elif score>=70:
        c+=1
    elif score>=60:
        d+=1
    else:
        f+=1

print("A:",a,"명")
print("B:",b,"명")
print("C:",c,"명")
print("D:",d,"명")
print("F:",f,"명")
```

```python
positive=0
negative=0
sum=0
for i in range(5):
    num = int(input(str(i+1)+"번째 숫자:"))

    if num>0:
        positive+=1
    elif num<0:
        negative+=1

    sum+=num

avg=sum/5
print("양수의 개수:",positive)
print("음수의 개수:",negative)
print("전체 합:",sum)
print("평균:",avg)

    
```

```python
positive=0
negative=0
sum=0

while True:
    num=int(input("숫자를 입력하세요:"))

    if num>0:
        print("양수입니다.")
        positive+=1
    elif num<0:
        print("음수입니다.")
        negative+=1
    else:
        break
    sum+=num

print("입력종료!")
print("양수:",positive,"개")
print("음수:",negative,"개")
print("전체 합:",sum)
```

```python
tot=0

while True:
    price=int(input("상품 가격을 입력하세요:"))

    if price==0:
        break
    tot+=price
if tot>=50000:
    disccunt_rate=0.2
elif tot>=30000:
    disccunt_rate=0.2
elif tot>=10000:
    disccunt_rate=0.05
else:
    disccunt_rate=0

disccunt=tot*disccunt_rate

print("결제결과")
print("할인전금액:",tot,"원")
print("할인율:",disccunt_rate*100,"%")
print("할인금액:",disccunt,"원")
```
